```python
import urllib.request as req
import json
import pandas as pd

def save_danmu_to_csv(sn):
    # url = "https://api.gamer.com.tw/anime/v1/danmu.php?videoSn={}&geo=TW%2CHK".format(sn)
    url = f"https://api.gamer.com.tw/anime/v1/danmu.php?videoSn={sn}&geo=TW%2CHK"
    f = req.urlopen(url)
    content = f.read()
    content = json.loads(content)

    danmus = content["data"]["danmu"]

    df = pd.DataFrame(danmus)
    df.to_csv("baha.csv", encoding="utf-8", index=False)
    return df

save_danmu_to_csv("26850")
```
