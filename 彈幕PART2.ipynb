````python
import urllib.request as req
import json
import pandas as pd
import os

def save_danmu_to_csv(sn, description=None):
    if description == None:
        dirname ="baha"
    else:
        dirname = f"baha/{description}"
    if not os.path.exists(dirname):
        os.makedirs(dirname)

    # url = "https://api.gamer.com.tw/anime/v1/danmu.php?videoSn={}&geo=TW%2CHK".format(sn)
    url = f"https://api.gamer.com.tw/anime/v1/danmu.php?videoSn={sn}&geo=TW%2CHK"
    f = req.urlopen(url)
    content = f.read()
    content = json.loads(content)

    danmus = content["data"]["danmu"]

    df = pd.DataFrame(danmus)
    df.to_csv(f"{dirname}/{sn}.csv",
        encoding="utf-8",
        index=False)
    return df

save_danmu_to_csv("26850", "鬼滅遊郭")
````

````python
import urllib.request as req
# BeautifulSoup: 幫忙分析HTML
import bs4 as bs
# 這裡被擋是因為巴哈擋國外IP，所以在COLAB上執行會被拒絕
# url = "https://ani.gamer.com.tw/animeVideo.php?sn=26850"
url = "https://github.com/Elwing-Chou/tibame_20260714/raw/refs/heads/main/baha.html"
h = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36"
}
r = req.Request(url, headers=h)
f = req.urlopen(r)
content = f.read()
# 型態轉換:
html = bs.BeautifulSoup(content)
# 專屬功能: find find_all(html名字, 篩選條件)
for link in html.find_all("a"):
    # 內容: .get_text()
    # 特殊屬性: ["href"]
    link_href = link["href"]
    if link_href.startswith("?sn"):
        print("下載彈幕", link_href)
        sn = link_href.split("=")[-1]
        save_danmu_to_csv(sn, description="鬼滅遊郭")
````

````python
import os
import glob
# glob可以幫忙找到符合需要的東西
import pandas
total = []

for fp in glob.glob("baha/*/*.csv"):
    df = pd.read_csv(fp, encoding="utf-8")
    sn = os.path.basename(fp).replace(".csv", "")
    # dataframe增加新行
    df["sn"] = sn
    total.append(df)

total_df = pd.concat(total)
# 如果你想reset列標籤 total_df = total_df.reset_index()
total_df
````
